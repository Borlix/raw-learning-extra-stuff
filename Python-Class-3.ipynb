```python
students = ["Peace", "Bob", "Charlie"]
students.append("Diano")
print(students)
students.insert(1, "Rose")
print(students)  
students.remove("Bob")
print(students)
last_student = students.pop()
print(last_student, "->",students)

print(len(students))
print("Peace"in students)
```
